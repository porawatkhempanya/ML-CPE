**Part 3: Data Cleaning**

**Perform**

```python
import pandas as pd

# Load Dataset
df = pd.read_csv("data.csv")

# -------------------------------
# 1. Display Missing Values
# -------------------------------
print("Missing Values:")
print(df.isnull().sum())

# -------------------------------
# 2. Fill Missing Values
# (Replace missing values with the mean)
# -------------------------------
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# -------------------------------
# 3. Display Duplicate Records
# -------------------------------
print("\nDuplicate Records:", df.duplicated().sum())

# -------------------------------
# 4. Remove Duplicate Records
# -------------------------------
df = df.drop_duplicates()

print("Shape After Removing Duplicates:", df.shape)

# -------------------------------
# 5. Check Data Types
# -------------------------------
print("\nData Types:")
print(df.dtypes)

# -------------------------------
# 6. Convert Data Types (if needed)
# -------------------------------
df["Duration"] = df["Duration"].astype(float)
df["Pulse"] = df["Pulse"].astype(float)
df["Maxpulse"] = df["Maxpulse"].astype(float)

print("\nData Types After Conversion:")
print(df.dtypes)

# -------------------------------
# 7. Save Cleaned Dataset
# -------------------------------
df.to_csv("cleaned_data.csv", index=False)

print("\nData Cleaning Completed Successfully!")
```

**Incorrect Data Correction**

```python
import pandas as pd

# Load cleaned dataset
df = pd.read_csv("data.csv")

print("Before Correction:")
print(df.describe())


# Check incorrect values

# Duration, Pulse, Maxpulse, Calories cannot be negative
df.loc[df["Duration"] < 0, "Duration"] = None
df.loc[df["Pulse"] < 0, "Pulse"] = None
df.loc[df["Maxpulse"] < 0, "Maxpulse"] = None
df.loc[df["Calories"] < 0, "Calories"] = None


# Correct missing values after correction
df["Duration"] = df["Duration"].fillna(df["Duration"].mean())
df["Pulse"] = df["Pulse"].fillna(df["Pulse"].mean())
df["Maxpulse"] = df["Maxpulse"].fillna(df["Maxpulse"].mean())
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())


print("\nAfter Correction:")
print(df.describe())


# Save corrected dataset
df.to_csv("data_corrected.csv", index=False)

print("\nIncorrect data has been corrected.")
```

**Part 4: Feature Engineering**

**Label Encoding**

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Example categorical data
df = pd.DataFrame({
    "Activity": [
        "Walking",
        "Running",
        "Cycling",
        "Walking",
        "Running"
    ]
})

print("Before Label Encoding:")
print(df)


# Create Label Encoder
encoder = LabelEncoder()

# Convert text to numbers
df["Activity_Encoded"] = encoder.fit_transform(df["Activity"])


print("\nAfter Label Encoding:")
print(df)


# Show mapping
print("\nLabel Mapping:")
for index, label in enumerate(encoder.classes_):
    print(label, "=", index)
```

**One-Hot Encoding**

```python
import pandas as pd

# Example categorical data
df = pd.DataFrame({
    "Activity": [
        "Walking",
        "Running",
        "Cycling",
        "Walking",
        "Running"
    ]
})

print("Before One-Hot Encoding:")
print(df)


# Apply One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=["Activity"])


print("\nAfter One-Hot Encoding:")
print(df_encoded)
```
